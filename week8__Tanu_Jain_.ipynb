## Tool 1 : Calculator

This tool evaluates mathematical expressions entered by the user.

If the expression is invalid, the tool returns an error message instead of crashing.

```python
# 🛠️ TOOL 1 : Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""

    try:
        return str(eval(expression))

    except Exception:
        return "Error in calculation"
```

## Tool 2 : Keyword Extractor

This tool extracts important words from a given text.

Working:
- Splits the text into words
- Converts words to lowercase
- Removes duplicate words
- Selects words having length greater than 4 characters
- Returns maximum 5 keywords

```python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
```

## Tool 3 : Word Counter
This is an additional tool added by me to count words in a text

```python
# 🛠️ TOOL 3: Word Counter
def word_count(text):
    return len(text.split())
```

## Designing the Agent

The agent performs three major steps:

Step 1:
Convert the query to lowercase for easy comparison.

Step 2:
Identify the user's intent.

Step 3:
Route the query to the appropriate tool.

Routing Conditions:

- If query contains "calculate" → Calculator Tool
- If query contains "keywords" → Keyword Extractor Tool
- If query contains "count words" → Word Counter Tool
- Otherwise → General Response Handler

The response is always returned in JSON format containing:
{
    "type": "...",
    "result": ...
}

```python
# 🤖 AGENT FUNCTION

def agent(query: str):
    try:
        query_lower = query.lower()
        print(f"[LOG] Received Query : {query}")

        # Calculator Routing
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()
            if expression == "":
                return {
                    "type": "error",
                    "result": "No mathematical expression found."
                }
            answer = calculator(expression)
            print("[LOG] Calculator Tool Selected")
            return {
                "type": "calculation",
                "result": answer
            }


        # Keyword Routing
        elif "keywords" in query_lower:
            text = query.replace("Extract keywords from", "")

            text = text.replace("keywords", "").strip()

            if text == "":
                return {
                    "type": "error",
                    "result": "No text provided for keyword extraction."
                }
            keywords = extract_keywords(text)
            print("[LOG] Keyword Tool Selected")
            return {
                "type": "keywords",
                "result": keywords
            }


        # Word Count Routing
        elif "count words" in query_lower:
                text = query_lower.replace("count words in", "").strip()
                return {
                     "type": "word_count",
                     "result": word_count(text)
               }


        # General Query Routing
        else:
            print("[LOG] General Response Selected")
            return {
                "type": "general",
                "result":
                f"I understand your query: {query}"
            }
    except Exception as e:
        return {
            "type": "error",
            "result": str(e)
        }
```

## Validation Testing

To verify the correctness of the agent, multiple test queries are executed.

The testing checks:
1. Proper intent detection
2. Correct tool selection
3. Structured JSON output
4. Error handling
5. Logging information

```python
# 🧪 Test Cases
queries = [
    "Calculate 20 + 5",
    "Calculate 120 / 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?",
    "count words in Hello",
    "Calculate",
    "Keywords"
]

for q in queries:
    print("Query :", q)
    response = agent(q)
    print(json.dumps(response,indent=4))
    print("*"*40)
```

## Interactive Mode

Interactive mode allows the user to continuously enter queries.

The agent keeps accepting input until the user types "exit".

This simulates a real-world AI assistant where users can ask multiple questions in one session.

```python
# 🎯 Interactive Mode

while True:
    user_input = input(
        "Enter query (type 'exit' to stop): "
    )
    if user_input.lower() == "exit":
        print("Assistant stopped.")
        break
    response = agent(user_input)
    print(json.dumps(response,indent=4))
```
